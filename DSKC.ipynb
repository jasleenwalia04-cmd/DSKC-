# TO FIND THE VALUE OF PI (2-D)

---



```python
import numpy as np
import matplotlib.pyplot as plt

# Number of points to generate
n = int(input("Enter the number of points you want to generate: "))

# Generate random points in the square [-1, 1] x [-1, 1]
x = np.random.uniform(-1, 1, n)
y = np.random.uniform(-1, 1, n)

# Check if points are inside the unit circle
inside_circle = (x**2 + y**2) <= 1

# Estimate pi (Area of unit circle is pi*r^2, here r=1 so area = pi)
# The fraction of points inside the circle to the square area gives pi/4
pi_estimate = (np.sum(inside_circle) / n) * 4

# Print the estimate
print(f"Estimated value of pi: {pi_estimate}")

# Plotting the points
plt.figure(figsize=(8, 8))
plt.scatter(x[inside_circle], y[inside_circle], color='blue', s=1, label='Inside Circle')
plt.scatter(x[~inside_circle], y[~inside_circle], color='red', s=1, label='Outside Circle')
plt.title('Estimating Pi Using Monte Carlo Method (2D)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

# TO CALCULATE THE VALUE OF PI (3-D)

```python

import numpy as np
import matplotlib.pyplot as plt

# Number of points to generate
n = int(input("Enter the number of points you want to generate: "))

# Generate random points
x = np.random.uniform(0, 1, n)
y = np.random.uniform(0, 1, n)
z = np.random.uniform(0, 1, n)

# Check if points are inside the unit sphere
inside_sphere = (x**2 + y**2 + z**2) <= 1

# Estimate pi (Volume of unit sphere is 4/3*pi*r^3, here r=1 so volume = 4/3*pi)
# The fraction of points in the first octant of the sphere to the cube volume gives (1/8)*(4/3)*pi
pi_estimate = (np.sum(inside_sphere) / n) * 6  # 8*(3/4) as the adjustment factor

# Print the estimate
print(f"Estimated value of pi: {pi_estimate}")

# Plotting the points (2D projection for visualization purposes)
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x[inside_sphere], y[inside_sphere], z[inside_sphere], color='blue', s=1, label='Inside Sphere')
ax.scatter(x[~inside_sphere], y[~inside_sphere], z[~inside_sphere], color='red', s=1, label='Outside Sphere')
ax.set_title('Estimating Pi Using Monte Carlo Method (3D)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.legend()
plt.show()
```

# TO PLOT THE GRAPH PI VS NUMBER OF DARTS TO EXSTIMATE THE VALUE OF PI

```python
import numpy as np
import matplotlib.pyplot as plt

def estimatePi(n):
  """Estimates pi using Monte Carlo simulation with n darts."""
  # Generate random points
  x = np.random.uniform(0, 1, n)
  y = np.random.uniform(0, 1, n)

  # Count points inside quarter circle
  inside = 0
  piEstimates = []

  # Iterate through points
  for i in range(n):
    if x[i]**2 + y[i]**2 <= 1:
      inside += 1
    piEstimate = 4 * inside / (i + 1)
    piEstimates.append(piEstimate)

  return piEstimates

# Number of darts for different cases
numDartsList = [1000, 5000, 10000]

# Set up the plot
plt.figure(figsize=(15, 10))

# Plot
for i, numDarts in enumerate(numDartsList):
  piEstimates = estimatePi(numDarts)

  # Subplot
  plt.subplot(3, 1, i + 1)
  plt.plot(piEstimates, label=f'Estimate with {numDarts} darts')
  plt.axhline(y=np.pi, color='r', linestyle='--', label='True value')
  plt.xlabel('Number of Darts')
  plt.ylabel('Estimate')
  plt.title(f'Estimating $\pi$ (Monte Carlo, {numDarts} darts)')
  plt.legend()

# Show the plots
plt.tight_layout()
plt.show()

```

# FINDING THE INTEGRAL BY HIT AND MISS MONTE CARLO METHOD OF THE FUNCTION SIN(x)

```python
import numpy as np
import matplotlib.pyplot as plt

# Defining the function
def f(x):
    return np.sin(x)

# Defining the intervals
a = 0
b = np.pi

# Finding the maximum value
def max(a, b):
    xv = np.linspace(a, b, 1000)
    maxidx = np.argmax(f(xv))
    max_x = xv[maxidx]
    max_y = f(max_x)
    return max_x, max_y

# Defining the number of darts
N = int(input("enter the number of points to be generated- "))

# Finding maximum x and y
max_x, max_y = max(a, b)

# Generating random points
xran = np.random.uniform(a, b, N)
yran = np.random.uniform(0, max_y, N)  # Use the correct max_y

# Counting points below the curve
nb = np.sum(yran <= f(xran))

# Rectangle area
rar = (b - a) * max_y

# Estimating
intg=(nb/N)*rar

#plotting the graph
x=np.linspace(a,b,500)
y=f(x)

plt.figure(figsize=(10,6))
plt.plot(x,y,label='f(x)',color="blue")
plt.fill_between(x,y,color="skyblue",alpha=0.4)

plt.scatter(xran,yran,color="red",s=1,label="random points")
plt.scatter(xran[yran<=f(xran)],yran[yran<=f(xran)],color="green",s=1,label="below curve")

plt.ylim(0,max_y*1.1)
plt.title("Monte Carlo Integration")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()

print("estimated integral value is ", intg)
print("Maximum value of function is ",max_y)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

# Data
z = np.array([0.07, 0.09, 0.12, 0.17, 0.1791, 0.1993, 0.2, 0.27, 0.28, 0.35, 0.3519, 0.3802, 0.4])
h_observed = np.array([69, 69, 68.6, 83, 75, 75, 72.9, 77, 88.8, 82.7, 83, 83, 95])
E = np.array([19.6, 12, 26.2, 8, 4, 5, 29.6, 14, 36.6, 8.4, 14, 13.5, 17])
h0 = 67.8  # Hubble constant

def model_function(z, Omega_m, h0):
    """
    Calculates the theoretical Hubble parameter (h) for a given redshift (z).
    """
    return h0 * np.sqrt(Omega_m * (1 + z) ** 3 + (1 - Omega_m))

def chi_square(Omega_m):
    """
    Calculates the chi-square value for a given Omega_m.
    """
    h_theoretical = model_function(z, Omega_m, h0)
    chi2 = np.sum(((h_observed - h_theoretical) / E) ** 2)
    return chi2

# Omega_m range (0 to 1 with 100 steps)
Omega_m_values = np.linspace(0, 1, 100)

# Calculate chi-square for each Omega_m value
chi_square_values = [chi_square(val) for val in Omega_m_values]

# Find the minimum chi-square value
min_chi_square = min(chi_square_values)

# Plot chi-square vs Omega_m
plt.figure(figsize=(10, 6))
plt.plot(Omega_m_values, chi_square_values, label='Chi-Square Curve')
plt.axhline(y=min_chi_square + 1, color='r', linestyle='--', label='1σ Confidence Level')
plt.axhline(y=min_chi_square + 4, color='g', linestyle='--', label='2σ Confidence Level')
plt.axhline(y=min_chi_square + 9, color='b', linestyle='--', label='3σ Confidence Level')

# Add labels and title
plt.xlabel('Omega_m')
plt.ylabel('Chi-Square')
plt.title('Chi-Square vs. Omega_m (Hubble Constant Data)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
from scipy.optimize import minimize
import matplotlib.pyplot as plt

# Data (use your actual data here)
z = np.array([0.07, 0.09, 0.12, 0.17, 0.1791, 0.1993, 0.2, 0.27, 0.28, 0.35, 0.3519, 0.3802, 0.4])
h_observed = np.array([69, 69, 68.6, 83, 75, 75, 72.9, 77, 88.8, 82.7, 83, 83, 95])
E = np.array([19.6, 12, 26.2, 8, 4, 5, 29.6, 14, 36.6, 8.4, 14, 13.5, 17])
h0 = 67.8  # Hubble constant

def model_function(z, params):
    Omega_m, Omega_L = params
    return h0 * np.sqrt(Omega_m * (1 + z)**3 + Omega_L)

def chi_square(params):
    h_theoretical = model_function(z, params)
    return np.sum(((h_observed - h_theoretical) / E)**2)

# Minimize chi-square function
initial_guess = [0.3, 0.7]
result = minimize(chi_square, initial_guess, method='Nelder-Mead')

print("Minimum chi-square value:", result.fun)
print("Optimal Omega_m:", result.x[0])
print("Optimal Omega_L:", result.x[1])

# Calculate chi-square for a range of Omega_m and Omega_L values
Omega_range = np.linspace(0, 1, 100)
chi_square_values = np.zeros((100, 100))

for i, om in enumerate(Omega_range):
    for j, ol in enumerate(Omega_range):
        chi_square_values[i, j] = chi_square([om, ol])

# Plotting
plt.figure(figsize=(10, 8))

# Calculate delta chi-square values for confidence intervals
chi_min = np.min(chi_square_values)
delta_chi_1sigma = 2.30  # For 2D, 68.3% confidence
delta_chi_2sigma = 6.17  # For 2D, 95.4% confidence
delta_chi_3sigma = 11.8  # For 2D, 99.7% confidence

# Plot contours
plt.contour(Omega_range, Omega_range, chi_square_values.T,
            levels=[chi_min + delta_chi_1sigma, chi_min + delta_chi_2sigma, chi_min + delta_chi_3sigma],
            colors=['black', 'red', 'green'], linestyles=['solid', 'solid', 'solid'])

plt.xlabel(r'$\Omega_m$', fontsize=18)
plt.ylabel(r'$\Omega_\Lambda$', fontsize=18)
plt.title('Parameter Estimation: Chi Square', fontsize=20)

# Mark the best fit point
plt.plot(result.x[0], result.x[1], 'k*', markersize=10, label='Best Fit')

plt.legend(loc="lower left")
plt.xlim(0, 1)
plt.ylim(0, 1)

# Add labels for confidence intervals
plt.text(0.8, 0.2, r'1$\sigma$', fontsize=12)
plt.text(0.85, 0.4, r'2$\sigma$', fontsize=12)
plt.text(0.9, 0.6, r'3$\sigma$', fontsize=12)

plt.savefig("omega_m_omega_l_contours.png")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
z=np.array([0.07  , 0.09  , 0.12  , 0.17  , 0.179 , 0.199 , 0.2   , 0.27  ,
       0.28  , 0.352 , 0.3802, 0.4   , 0.4004, 0.4247, 0.4497, 0.47  ,
       0.4783, 0.48  , 0.5929, 0.6797, 0.7812, 0.8754, 0.88  , 0.9   ,
       1.037 , 1.3   , 1.363 , 1.43  , 1.53  , 1.75  , 1.965 ]);


hz= np.array([ 69. ,  69. ,  68.6,  83. ,  75. ,  75. ,  72.9,  77. ,  88.8,
        83. ,  83. ,  95. ,  77. ,  87.1,  92.8,  89. ,  80.9,  97. ,
       104. ,  92. , 105. , 125. ,  90. , 117. , 154. , 168. , 160. ,
       177. , 140. , 202. , 186.5])

shz=np.array([19.6, 12. , 26.2,  8. ,  4. ,  5. , 29.6, 14. , 36.6, 14. , 13.5,
       17. , 10.2, 11.2, 12.9, 34. ,  9. , 62. , 13. ,  8. , 12. , 17. ,
       40. , 23. , 20. , 17. , 33.6, 18. , 14. , 40. , 50.4])

def log_likelihood(z,hz,shz,h0,om):
  n=len(z)
  sum=0
  for i in range(0,n):
    sum= sum+ (hz[i]-h0*(om*(1+z[i])**3+(1-om))**0.5)**2/(shz[i])**2
  return -(sum)

def log_prior1(h0,om):
  if 50< h0 < 100 and 0< om < 1:
        return 0.0
  return -np.inf

def log_prior2(h0,om):
  h0prior= -(h0-70)**2/(3**2)
  omprior= -(om-0.3)**2/(0.1**2)
  net= h0prior+omprior
  return net

def log_posterior(h0,om):
  func= log_likelihood(z,hz,shz,h0,om)+ log_prior1(h0,om)
  return func

import numpy as np
nchain=[]
achain=[]
bchain=[]
posteriorchain=[]
a0=70
b0=0.3                                 # initial value
s=1000000                       # no of steps
burnout_time= 1000                    # Burnout_time
t=0

for i in range(1,s):
  anew= a0+np.random.normal(0,0.1)
  bnew= b0+np.random.normal(0,0.01)
  u=np.random.uniform(0,1)
  alpha= np.exp(log_posterior(anew,bnew)- log_posterior(a0,b0))

  if log_posterior(anew,bnew)> log_posterior(a0,b0):
    t=t+1
    if t>=burnout_time:
      nchain.append(t)
      achain.append(anew)
      bchain.append(bnew)
      posteriorchain.append(log_posterior(a0,b0))
    a0=anew
    b0=bnew

  elif alpha >= u:
    t=t+1
    if t>=burnout_time:
      nchain.append(t)
      achain.append(anew)
      bchain.append(bnew)
      posteriorchain.append(log_posterior(a0,b0))
    a0=anew
    b0=bnew

  else:
    a0=a0
    b0=b0

plt.plot(nchain,achain)
ar= t/s
print("acceptance ratio=",ar)
```
